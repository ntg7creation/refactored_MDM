```python





plt.barh(y_pos, mse_values, color="skyblue", edgecolor="black")
plt.yticks(y_pos, scenes, fontsize=8)
plt.gca().invert_yaxis()
plt.xlabel("MSE Distance")
plt.title(f"Closest motions for {gen_file}")
for i, v in enumerate(mse_values):
    plt.text(v, i, f"{v:.4f}", va="center", ha="left", fontsize=7)
plt.tight_layout()
plt.show()
```
